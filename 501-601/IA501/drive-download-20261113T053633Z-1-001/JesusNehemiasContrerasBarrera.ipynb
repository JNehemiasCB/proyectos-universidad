```python
import time
import math

N = 1_000_000

lista1 = list(range(N))
lista2 = list(range(N))

start = time.time()
resultado = [lista1[i]+ lista2[i]for i in range(N)]
end = time.time()
print("python puro suma:",end - start, "segundo")

arr1 = np.arange(N)
arr2 = np.arange(N)

start = time.time()
resultado = arr1 + arr2
end = time.time()
print("Numpy suma:", end - start,"segundos ")

```
