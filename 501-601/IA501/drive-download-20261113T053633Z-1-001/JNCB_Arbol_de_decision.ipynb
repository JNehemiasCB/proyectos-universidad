```python
iris = load_iris()

iris_df = pd.DataFrame(
    data=iris.data,
    columns=['Sepal length (cm)', 'Sepal width (cm)', 'Petal length (cm)', 'Petal width (cm)']
)

iris_df.head()
```

```python
X = iris.data[:, [2, 3]]
y = iris.target
profundidad = 2
tamano_petalos = 1.0
X_modificado = X * tamano_petalos
feature_names = [iris.feature_names[i] for i in [2, 3]]

iris_df_sliced = iris_df.iloc[:, 2:]
pd.DataFrame(
    data=np.column_stack([iris_df_sliced, y]),
    columns=['Petal length (cm)', 'Petal width (cm)', 'Result']
)
```

```python
clf = DecisionTreeClassifier(max_depth=profundidad, random_state=42)
clf.fit(X_modificado, y)
```

```python
clf.predict([[2.85, 1.2]])
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(12, 8))
plot_tree(
    clf,
    feature_names=["petal length (cm)", "petal width (cm)"],
    class_names=iris.target_names,
    filled=True,
    rounded=True,
    fontsize=12
)

plt.title(f"Árbol con profundidad={profundidad} y tamaño de pétalos x(tamaño_pétalos)")
plt.show()
```
