Lectura de datos tabulares desde archivos CSV  o Excel

```
import pandas as pd

# Leer todas las columnas
df_csv = pd.read_csv("/content/dataset_pandas.csv")
df_csv
```

```
#Leer solo ciertas columnas
df_csv = pd.read_csv("/content/dataset_pandas.csv", usecols=['edad','email','estado'])
df_csv
```

```
#Leer todas las columnas pero NO leer el footer
df_csv = pd.read_csv("/content/dataset_pandas.csv", skipfooter=2)
df_csv
```

```
# Leer excel (por defecto se lee la primera hoja)
df_xlsx = pd.read_excel("/content/dataset_pandas.xlsx")
df_xlsx
```

```
#Leer excel (por defecto se lee la primera hoja)
df_xlsx = pd.read_excel("/content/dataset_pandas.xlsx", sheet_name="mini_tabla")
df_xlsx
```

4.2 Extraer un rango de filas

4.3 Filtrando filas y columnas

```
df_csv.loc[df_csv['edad']>35, ['email', 'IMC']]
```

```
df_csv.sort_values (by= 'edad', ascending= False)
```

Ordenamos de  manera ascendente por edad e IMC

```
df_csv.sort_values (by= ['edad', 'IMC'])
```

Ahora por estados (alfabeticamente)

```
df_csv.sort_values (by= ['estado'], ascending=False)
```

```
df_csv.rename (columns={'edad' : 'Edad', 'genero':'género'})
```

Eliminar columnas

```
df_csv.drop(columns=['email'], inplace=True)
```

```
df_csv.groupby(['estado'])['IMC'].mean()
```

Podemos organizar lo anterior de forma descendente, por ejemplo por IMC

```
df_imc = df_csv.groupby(['estado'])['IMC'].mean().sort_values(ascending=False)
df_imc
```

Almacenamiento de un dataframe

```
df_imc.to_csv('/content/datos_IMC.csv')
```
