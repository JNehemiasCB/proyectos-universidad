```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (8,6)
plt.rcParams['axes.grid'] = True
```

```python
niveles = np.array([5,10,15,20], dtype=float)
daño = np.array([12,20,30,40], dtype=float)
nombres = ['pikachu', 'charmander', 'bulbasaur', 'squirtle']

pd.DataFrame({'pokemon':nombres, 'Nivel':niveles, 'daño':daño})
```

```python
niveles = np.array([5,10,15,20], dtype=float)
daño = np.array([12,20,30,40], dtype=float)
nombres = ['pikachu', 'charmander', 'bulbasaur', 'squirtle']

pd.DataFrame({'pokemon':nombres, 'Nivel':niveles, 'daño':daño})
```

```python
plt.figure()
plt.scatter(niveles, daño, s=200)
for x, y, name in zip(niveles,daño, nombres):
  plt.text(x + 0.4, y + 0.6, name)
plt.title('pokemon: Nivel vs daño')
plt.xlabel('nivel (x)')
plt.ylabel('daño (y)')
plt.show()
```

```python
m, b = np.polyfit(niveles, daño, deg=1)
m, b
```

```python
x_linea = np.linspace(0, 25, 100)
y_linea = m * x_linea + b

plt.figure()
plt.scatter(niveles, daño, s=200, label='Datos')
plt.plot(x_linea, y_linea, linestyle='--', label=f'Linea: y = {m: .2f}x + {b: .2f}')
for x, y, name in zip(niveles,daño, nombres):
  plt.text(x + 0.4, y + 0.6, name)
  plt.title('Linea de regresion: Nivel vs Daño')
  plt.xlabel('nivel (x)')
  plt.ylabel('daño (y)')
  plt.legend()
  plt.show()
```

```python
nivel_eevee = 25.0
daño_pred_Eevee = m * nivel_eevee + b
daño_pred_Eevee
```

```python
daño_pred = m * niveles + b
error = daño - daño_pred

tabla = pd.DataFrame({
    'pokemon':nombres,
    'Nivel':niveles,
    'daño real':daño,
    'daño_predicho': np.round(daño_pred, 2),
    'error (Real - Pred)': np.round(error, 2)
})
tabla.loc[len(tabla)] = ['eevee (predicho)', 25.0, np.nan, np.round(daño_pred_Eevee, 2), np.nan]
tabla
```
